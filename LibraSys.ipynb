```python
# ==========================================
#      LIBRARY MANAGEMENT SYSTEM
# ==========================================

books = []
students = []


# ==========================================
# 1. ADD BOOK
# ==========================================

def add_book():
    print("\n---------- ADD BOOK ----------")

    book_id = int(input("Enter Book ID: "))

    # Check if book ID already exists
    for book in books:
        if book["id"] == book_id:
            print("Book ID already exists!")
            return

    name = input("Enter Book Name: ")
    author = input("Enter Author Name: ")

    book = {
        "id": book_id,
        "name": name,
        "author": author,
        "status": "Available",
        "issued_to": "None"
    }

    books.append(book)

    print("Book added successfully!")


# ==========================================
# 2. VIEW ALL BOOKS
# ==========================================

def view_books():
    print("\n---------- ALL BOOKS ----------")

    if len(books) == 0:
        print("No books available.")
        return

    for book in books:
        print("\nBook ID     :", book["id"])
        print("Book Name   :", book["name"])
        print("Author      :", book["author"])
        print("Status      :", book["status"])
        print("Issued To   :", book["issued_to"])
        print("-------------------------------")


# ==========================================
# 3. SEARCH BOOK BY ID
# ==========================================

def search_by_id():
    print("\n---------- SEARCH BOOK ----------")

    book_id = int(input("Enter Book ID: "))

    for book in books:
        if book["id"] == book_id:
            print("\nBook Found!")
            print("Book ID   :", book["id"])
            print("Name      :", book["name"])
            print("Author    :", book["author"])
            print("Status    :", book["status"])
            print("Issued To :", book["issued_to"])
            return

    print("Book not found.")


# ==========================================
# 4. SEARCH BOOK BY NAME
# ==========================================

def search_by_name():
    print("\n---------- SEARCH BOOK ----------")

    name = input("Enter Book Name: ")

    found = False

    for book in books:
        if book["name"].lower() == name.lower():
            print("\nBook Found!")
            print("Book ID   :", book["id"])
            print("Name      :", book["name"])
            print("Author    :", book["author"])
            print("Status    :", book["status"])
            found = True

    if found == False:
        print("Book not found.")


# ==========================================
# 5. ISSUE BOOK
# ==========================================

def issue_book():
    print("\n---------- ISSUE BOOK ----------")

    book_id = int(input("Enter Book ID: "))

    for book in books:

        if book["id"] == book_id:

            if book["status"] == "Issued":
                print("Book is already issued.")
                return

            student_id = int(input("Enter Student ID: "))

            # Check whether student exists
            for student in students:
                if student["id"] == student_id:

                    book["status"] = "Issued"
                    book["issued_to"] = student["name"]

                    print("Book issued successfully!")
                    return

            print("Student not found.")
            return

    print("Book not found.")


# ==========================================
# 6. RETURN BOOK
# ==========================================

def return_book():
    print("\n---------- RETURN BOOK ----------")

    book_id = int(input("Enter Book ID: "))

    for book in books:

        if book["id"] == book_id:

            if book["status"] == "Available":
                print("This book is not issued.")
                return

            book["status"] = "Available"
            book["issued_to"] = "None"

            print("Book returned successfully!")
            return

    print("Book not found.")


# ==========================================
# 7. REMOVE BOOK
# ==========================================

def remove_book():
    print("\n---------- REMOVE BOOK ----------")

    book_id = int(input("Enter Book ID: "))

    for book in books:

        if book["id"] == book_id:

            if book["status"] == "Issued":
                print("Issued books cannot be removed.")
                return

            books.remove(book)

            print("Book removed successfully!")
            return

    print("Book not found.")


# ==========================================
# 8. UPDATE BOOK
# ==========================================

def update_book():
    print("\n---------- UPDATE BOOK ----------")

    book_id = int(input("Enter Book ID: "))

    for book in books:

        if book["id"] == book_id:

            print("Current Name  :", book["name"])
            print("Current Author:", book["author"])

            new_name = input("Enter New Book Name: ")
            new_author = input("Enter New Author Name: ")

            book["name"] = new_name
            book["author"] = new_author

            print("Book details updated!")
            return

    print("Book not found.")


# ==========================================
# 9. SHOW AVAILABLE BOOKS
# ==========================================

def available_books():
    print("\n---------- AVAILABLE BOOKS ----------")

    found = False

    for book in books:

        if book["status"] == "Available":

            print("ID:", book["id"],
                  "| Name:", book["name"],
                  "| Author:", book["author"])

            found = True

    if found == False:
        print("No books are currently available.")


# ==========================================
# 10. SHOW ISSUED BOOKS
# ==========================================

def issued_books():
    print("\n---------- ISSUED BOOKS ----------")

    found = False

    for book in books:

        if book["status"] == "Issued":

            print("ID:", book["id"])
            print("Name:", book["name"])
            print("Author:", book["author"])
            print("Issued To:", book["issued_to"])
            print("-------------------------")

            found = True

    if found == False:
        print("No books are currently issued.")


# ==========================================
# 11. LIBRARY STATISTICS
# ==========================================

def statistics():
    print("\n---------- LIBRARY STATISTICS ----------")

    total_books = len(books)
    available = 0
    issued = 0

    for book in books:

        if book["status"] == "Available":
            available += 1

        elif book["status"] == "Issued":
            issued += 1

    print("Total Books     :", total_books)
    print("Available Books :", available)
    print("Issued Books    :", issued)
    print("Total Students  :", len(students))


# ==========================================
# 12. ADD STUDENT
# ==========================================

def add_student():
    print("\n---------- ADD STUDENT ----------")

    student_id = int(input("Enter Student ID: "))

    for student in students:

        if student["id"] == student_id:
            print("Student ID already exists!")
            return

    name = input("Enter Student Name: ")
    course = input("Enter Course: ")

    student = {
        "id": student_id,
        "name": name,
        "course": course
    }

    students.append(student)

    print("Student added successfully!")


# ==========================================
# 13. VIEW STUDENTS
# ==========================================

def view_students():
    print("\n---------- ALL STUDENTS ----------")

    if len(students) == 0:
        print("No students registered.")
        return

    for student in students:

        print("\nStudent ID :", student["id"])
        print("Name       :", student["name"])
        print("Course     :", student["course"])
        print("---------------------------")


# ==========================================
# 14. MAIN MENU
# ==========================================

while True:

    print("\n")
    print("========================================")
    print("       LIBRARY MANAGEMENT SYSTEM")
    print("========================================")

    print("1. Add Book")
    print("2. View All Books")
    print("3. Search Book by ID")
    print("4. Search Book by Name")
    print("5. Issue Book")
    print("6. Return Book")
    print("7. Remove Book")
    print("8. Update Book")
    print("9. Show Available Books")
    print("10. Show Issued Books")
    print("11. Library Statistics")
    print("12. Add Student")
    print("13. View Students")
    print("14. Exit")

    choice = input("\nEnter your choice: ")


    if choice == "1":
        add_book()

    elif choice == "2":
        view_books()

    elif choice == "3":
        search_by_id()

    elif choice == "4":
        search_by_name()

    elif choice == "5":
        issue_book()

    elif choice == "6":
        return_book()

    elif choice == "7":
        remove_book()

    elif choice == "8":
        update_book()

    elif choice == "9":
        available_books()

    elif choice == "10":
        issued_books()

    elif choice == "11":
        statistics()

    elif choice == "12":
        add_student()

    elif choice == "13":
        view_students()

    elif choice == "14":
        print("\nThank you for using the Library Management System!")
        break

    else:
        print("Invalid choice. Please try again.")
```
